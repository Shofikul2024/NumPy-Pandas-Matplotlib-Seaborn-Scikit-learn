##  (NUMPY)নামপাই কি?

পাইথনে নিউমেরিক্যাল ক্যালকুলেশনের  জন্য দারুণ একটি প্যাকেজ হচ্ছে NumPy। যার পূর্ণরুপ হচ্ছে Numarical Python।  নামপাইতে লিনিয়ার অ্যালজেব্রা, ফুরিয়ার ট্রান্সফরমেশন, মেট্রিক্স সহ অন্যান্য গাণিতিক ফাংশন গুলো রয়েছে। NumPy নিয়ে কাজ করাও দারুণ সহজ। বর্তমানে বেশির ভাগ মেশিন লার্নিং লাইব্রেরীতে নামপাই ব্যবহৃত হয়েছে।

পাইথনে বিল্টইন অ্যারে নেই। লিস্ট দিয়ে যদিও অ্যারের কাজ করা যায়, তবে অ্যারে থেকে স্লো। নামপাই ব্যবহার করে পাইথনে অ্যারে নিয়ে কাজ করা যায়। যা লিস্ট থেকে প্রায় ৫০ গুণ দ্রুত কাজ করে। নামপাই অ্যারে অবজেক্টকে বলে ndarray। ডেটা সাইন্স, মেশিন লার্নিং এর ক্ষেত্রে বেশির ভাগ সময় অ্যারে নিয়ে কাজ করতে হয়। আর যেখানে স্পিড একটা গুরুত্বপূর্ণ ফ্যাক্টর হিসেবে কাজ করে। নামপাই অ্যারে সি অ্যারের মত। যেখানে ডেটা গুলো কম্পিউটার মেমরিতে সিরিয়ালি রাখা হয়।  যার কারণে 
দ্রুততার সাথে ডেটা এক্সেস এবং কম্পিউট করা যায়। এমনকি নামপাই এর বেশির ভাগ অংশ সি প্রোগ্রামিং এ লেখা।

# NumPy Array Slicing :

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5, 6, 7])

print(arr[1:5]) 
```

Use the step value to determine the step of the slicing:

```python
# NumPy Array Slicing :
import numpy as np

arr = np.array([1, 2, 3, 4, 5, 6, 7])

print(arr[1:5:2])  # start=1, stop=5, step=2        step 2 mean  ekta skip kore nibe   index= value 1 theke suru kore 5 er age porjonto 2 step kore nibe, so  start value 2, then skip 3, then 4   so output hobe 2,4
```

```python
# NumPy Array Slicing :
import numpy as np

arr = np.array([1, 2, 3, 4, 5, 6, 7])  

print(arr[::2]) # start index 0  ,   end value  steps of 2
```

এখানে দুইটা 3*4 অ্যারে রয়েছে। 
বুঝতে আরেকটু সহজ হবে যদি আমরা এই অ্যারেটি জেনারেট করিঃ

```python

 
import numpy as np                  # create a 3D array of zeros
 
three_d = np.zeros((2, 3, 4))       # 2 blocks, 3 rows, 4 columns
 
print(three_d)                     # prints the 3D array three_d                        
```

প্রথম অ্যারেটাই আমরা 0 দিয়ে তৈরি করেছি। যদি (2, 3, 4) এর পরিবর্তে (4, 3, 4) লিখি, তাহলেঃ

মানে হচ্ছে 4টা 3*4 অ্যারে।
 যখন এমন মাল্টি ডাইমেনশন অ্যারে নিয়ে কাজ করব,
  তখন উল্টা থেকে পড়ব। যেমন (2, 3, 4) কে পড়ব প্রতিটা রোতে ৪টা আইটেম রয়েছে। 
  চারটা আইটেম করে ৩টা কলাম রয়েছে প্রতি স্লাইচে। এবং স্লাইচ রয়েছে ২টা।

```python
import numpy as np                # create a 3D array of zeros
 
three_d = np.zeros((4, 3, 4))     # 4 blocks, 3 rows, 4 columns
 
print(three_d)                   # prints the 3D array three_d
```

আমার মত রো এবং কলাম কোনটা, ভুলে গেলে আবার স্মরণ করে নিন। প্রথমে রো, পরে কলাম, পরে স্লাইস। তাহলে (4, 3, 4) এর ক্ষেত্রে হবে প্রতি রোতে ৪টা আইটেম করে মোট ৩টা রো রয়েছে, আর স্লাইস রয়েছে ৪টা! আবার অ্যারেতে কত গুলো আইটেম রয়েছে টোটাল তা এই (4, 3, 4) সংখ্যা গুলো গুণ করলেই পাওয়া যাবে। অর্থাৎ (4, 3, 4) সাইজের অ্যারেতে 48টা আইটেম থাকবে।

এবার একটা 3D অ্যারে র‍্যান্ডম নাম্বার দিয়ে তৈরি করিঃ
