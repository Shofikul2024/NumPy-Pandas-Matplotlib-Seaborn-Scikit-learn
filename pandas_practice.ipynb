 Pandas হল Python-এর একটি অত্যন্ত জনপ্রিয় এবং শক্তিশালী লাইব্রেরি, যা ডেটা ম্যানিপুলেশন এবং বিশ্লেষণের জন্য ব্যবহৃত হয়। এটি মূলত DataFrame এবং Series নামক ডেটা স্ট্রাকচার ব্যবহার করে ডেটাকে প্রক্রিয়া এবং বিশ্লেষণ করার জন্য তৈরি হয়েছে। Pandas ডেটা বিশ্লেষণকে সহজ এবং দ্রুত করে তোলে, এবং এটি ডেটাবেস, CSV, Excel, SQL এবং অন্যান্য ফাইল ফরম্যাট থেকে ডেটা আমদানি এবং রপ্তানি করার জন্য শক্তিশালী টুলস প্রদান করে।

Pandas এর প্রধান বৈশিষ্ট্য:
DataFrame: এটি একটি টেবিল আকারে ডেটা স্ট্রাকচার, যা সারি এবং কলাম সহ ডেটাকে সঞ্চালন ও বিশ্লেষণ করতে সহায়ক।
Series: এটি একটি একক কলাম বিশিষ্ট ডেটা স্ট্রাকচার, যা মূলত একক ডেটা সিরিজ ধারণ করতে ব্যবহৃত হয়।
ডেটা ক্লিনিং: মিসিং ডেটা পূর্ণ করা, ডেটা ফিল্টারিং, ডেটা কনভার্সন, এবং অপ্রয়োজনীয় ডেটা অপসারণের জন্য Pandas অত্যন্ত কার্যকর।
গ্রুপবাই অপারেশন: একসাথে ডেটা গ্রুপ করা এবং বিভিন্ন পরিসংখ্যান অপারেশন সম্পাদন করা যায়।
টাইম সিরিজ: Pandas টাইম সিরিজ ডেটার সাথে কাজ করতে সক্ষম, যা আর্থিক এবং বৈজ্ঞানিক বিশ্লেষণে অত্যন্ত সহায়ক।
ডেটার ফরম্যাটিং এবং ম্যানিপুলেশন: ডেটা ট্রান্সফরমেশন, সোর্টিং, এবং কমপ্লেক্স ক্যালকুলেশন করতে সহায়ক।

```python

 
import pandas as pd
 
data = {"Course ID": ['CSE10', 'CSE20', 'CSE30', 'CSE40'],
        "GPA": [4, 3.68, 3.55,3.98]
}
 
pd_data = pd.DataFrame(data)
 
print(pd_data)
```

```python
# নির্দিষ্ট কোন সারিকে অ্যাক্সেস করা
# পূর্বে আমরা জেনেছি যে পানডাস এর সিরিজ হচ্ছে এক একটি কলাম; কিন্তু এমন যদি হয় যে একটি সারি নিয়ে কাজ করতে হবে তখন আমরা loc() ব্যবহার করতে পারি। উপরের উদাহরণটিকে ব্যবহার করে বিষয়টি বুঝে নিই-

 
import pandas as pd
 
data = {"Course ID": ['CSE10', 'CSE20', 'CSE30', 'CSE40'],
        "GPA": [4, 3.68, 3.55,3.98]
}
 
pd_data = pd.DataFrame(data)
 
print(pd_data.loc[1])
```

```python
# CSV ফাইল নিয়ে কাজ করা
# আমি ধরে নিচ্ছি যে আপনাদের কম্পিউটার বা সিস্টেম এ কোন .csv ফাইল নেই, তাই আমরা একটি .csv ফাইল তৈরি করে নেব। বোঝার সুবধার্থে উপরের ডাটাফ্রেমটিকেই ব্যবহার করছি।

 
import pandas as pd
 
data = {"Course ID": ['CSE10', 'CSE20', 'CSE30', 'CSE40'],
        "GPA": [4, 3.68, 3.55,3.98]
}
pd_data = pd.DataFrame(data)
 
pd_data.to_csv('DataFrame.csv', index=False)



# এই প্রোগ্রামটি রান করলেই দেখতে পাবেন যে আপনার ফাইল ডিরেকটরি তে DataFrame নামে একটি .csv ফাইল তৈরি হয়ে গেছে। এই ফাইলটি খুললে দেখতে পাবেন যে এটার ভিতর pd_data এর ভ্যালুগুলোই আছে। এখানে to_csv মেথডটি দিয়ে পানডাস ডাটাফ্রেমকে csv ফাইলে রুপান্তর করা হয়েছে। index কে false করা হয়েছে কারণ আমরা চাইনা যে index এর ভ্যালুও ফাইলে সেভ হোক। আপনারা এটি বাদ দিয়েও চেষ্টা করে দেখতে পারেন।

# যেহেতু আমাদের ফাইল ডিরেক্টরিতে একটি .csv ফাইল আছেই তাই আমরা খুব সহজেই সেই ফাইলটিকে পড়তে পারব। এটার জন্য read_csv() মেথডটি ব্যবহার করব।
```

```python
# লুপ
# পানডাস ডাটাফ্রেম এর ভ্যালু গুলিকে লুপ দিয়ে এক্সেস করা যায়, সেক্ষেত্রে ইচ্ছামত যেকোন কন্ডিশনাল স্টেটমেন্ট দিয়ে কাঙ্খিত ফলাফল পাওয়া যায়।

 
import pandas as pd
 
pd_data = pd.read_csv('DataFrame.csv')
 
for x in pd_data.index:
  if pd_data.loc[x, "GPA"] > 3.9:
    pd_data.loc[x, "GPA"] = 4
 
print(pd_data)
```

```python
# ১. DataFrame তৈরি

import pandas as pd

# ডেটা সেট তৈরি
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Edward'],
    'Age': [24, 27, 22, 32, 29],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix']
}

df = pd.DataFrame(data)

# DataFrame দেখানো
print(df)
```

```python
# প্রথম সারি নির্বাচন
first_row = df.iloc[0]
print(first_row)
```

```python
# প্রথম সারি নির্বাচন
first_row = df.loc[0]
print(first_row)
```

```python
# Quick examples showing .loc (label-based) vs .iloc (position-based)
print("Index labels:", df.index.tolist())
print("\nDataFrame:\n", df)

# 1) Single-row access
print("\n--- Single-row ---")
print("df.loc[0] (label 0):\n", df.loc[0])
print("df.iloc[0] (position 0):\n", df.iloc[0])

# 2) Accessing a single cell
print("\n--- Single cell ---")
print("df.loc[0, 'Age'] ->", df.loc[0, 'Age'])
print("df.iloc[0, 1] ->", df.iloc[0, 1])

# 3) Slicing difference: loc end is inclusive, iloc end is exclusive
print("\n--- Slicing ---")
print("df.loc[0:2] (includes label 2):\n", df.loc[0:2])
print("df.iloc[0:2] (excludes position 2):\n", df.iloc[0:2])

# 4) Boolean indexing works naturally with loc
print("\n--- Boolean indexing with loc ---")
print(df.loc[df['Age'] > 25])

# 5) When index labels are integers, loc treats them as labels while iloc uses positions
# (uses existing variable x from the notebook)
print(f"\n--- Label vs position for x={x} ---")
print("df.loc[x]:\n", df.loc[x])
print("df.iloc[x]:\n", df.iloc[x])
```

```python
# ৩. groupby() ফাংশনের ব্যবহার

import pandas as pd

# উদাহরণ ডেটা তৈরি
data = {
    'Department': ['HR', 'IT', 'HR', 'IT', 'Sales', 'Sales', 'IT', 'HR'],
    'Employee': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank', 'Grace', 'Hank'],
    'Sales': [200, 300, 150, 400, 500, 600, 350, 250]
}

df = pd.DataFrame(data)

# 'Department' অনুযায়ী ডেটা গ্রুপ করা এবং গড় বিক্রয় বের করা
grouped = df.groupby('Department')['Sales'].mean()

print(grouped)
```
